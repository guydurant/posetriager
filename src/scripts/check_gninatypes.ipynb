```python
non_gninatypes_csv = pd.read_csv("/vols/opig/projects/durant-crossdocked/datasets/csv_files/redocked_train.csv")
gninatypes_csv = pd.read_csv("/vols/opig/projects/durant-crossdocked/datasets/csv_files/redocked_train_gninatypes.csv")
```

```python
def process_row(row, base_path, n_features=11, edge_radius=10):
    parser = StructuralFileParser()
    lig_path = Path(base_path) / row["ligand"]
    rec_path = Path(base_path) / row["protein"]
    lig_num = int(row.get("ligand_num", 0))
    label = row.get("label", 0)

    try:
        rec_df = parser.rdkitmol_to_df(rec_path, mol_type="receptor")
        rec_df = rec_df[rec_df["types"] > 1]  # Filter out non-atom rows
        # print(rec_df)
        lig_df = parser.rdkitmol_to_df(lig_path, mol_type="ligand", num=lig_num)
        lig_df = lig_df[lig_df["types"] > 1]  # Filter out non-atom rows
        print(lig_df)
        struct = make_box(
            concat_structs(rec_df, lig_df, n_features),
            radius=6,
            relative_to_ligand=True,
        )
        struct = struct[struct["atomic_number"] > 1]
        pos = torch.tensor(np.vstack([struct["x"], struct["y"], struct["z"]]).T, dtype=torch.float)
        x = make_bit_vector(struct.types.to_numpy(), n_features, compact=True)

        struct, edge_indices, edge_attrs = generate_edges(
            struct, inter_radius=edge_radius, intra_radius=2.0
        )
        edge_index = torch.tensor(np.vstack(edge_indices), dtype=torch.long)
        edge_attr = one_hot(torch.tensor(edge_attrs, dtype=torch.long), 3)

        y = torch.tensor(label)
        data = Data(x=x.float(), edge_index=edge_index, edge_attr=edge_attr, pos=pos, y=y)

    except Exception as e:
        warnings.warn(f"Error processing row {row['id']}: {e}")
        return None

    return data
```

```python
from rdkit import Chem
from pathlib import Path
    

def read_rdkit_file(path, base_path):
    path = Path(base_path) / path
    mol = Chem.MolFromMolFile(str(path))
    if mol is None:
        raise ValueError(f"Could not read RDKit file: {path}")
    return mol

success = 0
count = 0
for non_gninatypes_row in tqdm(non_gninatypes_csv.itertuples(), total=len(non_gninatypes_csv)):
    count += 1
    try:
        non_gninatypes_data = read_rdkit_file(non_gninatypes_row.ligand, "/data/localhost/not-backed-up/durant/CrossDocked2020/")
        if non_gninatypes_data is not None:
            success += 1
            # print(f"Processed {non_gninatypes_row.ligand} successfully.")
    except Exception as e:
        print(f"Failed to process {non_gninatypes_row.ligand}: {e}")
    if count > 300000:
        break

print(f"Successfully processed {success} out of {count} non-GNINATypes ligands.")
```
